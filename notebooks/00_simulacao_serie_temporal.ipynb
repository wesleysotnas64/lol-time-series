```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Reprodutibilidade dos resultados
np.random.seed(42)

# Número de observações
n = 365

# Índice temporal diário
datas = pd.date_range(
    start="2025-01-01",
    periods=n,
    freq="D"
)

# Componentes da série temporal
tendencia = np.linspace(10, 30, n)

sazonalidade = 5 * np.sin(
    2 * np.pi * np.arange(n) / 30
)

ruido = np.random.normal(
    loc=0,
    scale=2,
    size=n
)

# Série temporal simulada
valores = tendencia + sazonalidade + ruido

# Criar DataFrame
df = pd.DataFrame({
    "data": datas,
    "valor": valores
})

df.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["data"],
    df["valor"],
    label="Série temporal artificial"
)

plt.title("Simulação de uma Série Temporal")
plt.xlabel("Data")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
