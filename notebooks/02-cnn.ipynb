# 02 - CNN-1D

## Objetivo

Desenvolver uma Rede Neural Convolucional 1D (CNN-1D) para classificar o vencedor de partidas de League of Legends, utilizando as informações dos primeiros 20 minutos de cada partida.

## Dados

- 10.000 partidas;
- 21 instantes temporais (minutos 0 a 20);
- 105 características por instante;
- Formato de entrada: `(10000, 21, 105)`;
- Classificação binária: equipe azul (100) ou vermelha (200).

## Etapas

1. Preparação e organização dos dados.
2. Divisão cronológica em treinamento, validação e teste (70/15/15).
3. Normalização das características.
4. Construção e treinamento da CNN-1D.
5. Avaliação utilizando Accuracy, F1 Macro e Balanced Accuracy.
6. Comparação com os modelos baseline, Random Forest e XGBoost.


```python
import pandas as pd

# Configuração para reprodutibilidade
SEED = 42

# Carregamento do dataset
DATASET_PATH = "../data/team-time-series-10000.xlsx"

df = pd.read_excel(DATASET_PATH)

print(f"Seed definida: {SEED}")
print(f"Dimensões do dataset: {df.shape[0]} linhas e {df.shape[1]} colunas")

display(df.head())
```

```python
# Filtragem dos primeiros 20 minutos (inclusive)
df_filtrado = df[df["timestep"].between(0, 20)].copy()

print(f"Dimensões após filtragem: {df_filtrado.shape}")
print(f"Partidas encontradas: {df_filtrado['match_id'].nunique()}")
print(f"Timesteps encontrados: {sorted(df_filtrado['timestep'].unique())}")

display(df_filtrado.head())
```

```python
# Identificação das colunas de entrada
feature_cols = [
    col for col in df_filtrado.columns
    if col.startswith(("blue_", "red_", "diff_"))
]

# Separação das características e do vencedor
X = df_filtrado[feature_cols]
y = df_filtrado["vencedor"]

print(f"Quantidade de características: {len(feature_cols)}")
print(f"Dimensões de X: {X.shape}")
print(f"Dimensões de y: {y.shape}")

print("\nDistribuição dos vencedores:")
print(y.value_counts())
```

```python
# Verificação da quantidade de registros por partida
registros_por_partida = df_filtrado.groupby("match_id").size()

# Verificação da consistência do vencedor
vencedores_por_partida = df_filtrado.groupby("match_id")["vencedor"].nunique()

print(f"Partidas com exatamente 21 registros: {(registros_por_partida == 21).sum()}")
print(f"Partidas com exatamente um vencedor: {(vencedores_por_partida == 1).sum()}")

print(f"Menor quantidade de registros: {registros_por_partida.min()}")
print(f"Maior quantidade de registros: {registros_por_partida.max()}")
```

```python
import numpy as np

# Ordenação cronológica dos registros
df_filtrado = df_filtrado.sort_values(
    by=["match_id", "timestep"]
).reset_index(drop=True)

# Recuperação das características após a ordenação
X = df_filtrado[feature_cols].to_numpy()

# Transformação: partidas × instantes × características
X = X.reshape(10000, 21, 105)

# Um único vencedor por partida
y = df_filtrado.groupby("match_id", sort=False)["vencedor"].first().to_numpy()

print(f"Dimensões de X: {X.shape}")
print(f"Dimensões de y: {y.shape}")
```

```python
# Índices para divisão cronológica
n = len(X)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

# Separação dos conjuntos
X_train, y_train = X[:train_end], y[:train_end]

X_val, y_val = X[train_end:val_end], y[train_end:val_end]

X_test, y_test = X[val_end:], y[val_end:]

print("Treinamento:", X_train.shape, y_train.shape)
print("Validação:", X_val.shape, y_val.shape)
print("Teste:", X_test.shape, y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Inicialização do normalizador
scaler = StandardScaler()

# Transformação temporária para duas dimensões
X_train_2d = X_train.reshape(-1, 105)

# Ajuste e normalização do treinamento
X_train_scaled = scaler.fit_transform(X_train_2d)

# Normalização da validação e do teste usando os parâmetros do treinamento
X_val_scaled = scaler.transform(X_val.reshape(-1, 105))
X_test_scaled = scaler.transform(X_test.reshape(-1, 105))

# Recuperação do formato tridimensional
X_train = X_train_scaled.reshape(7000, 21, 105)
X_val = X_val_scaled.reshape(1500, 21, 105)
X_test = X_test_scaled.reshape(1500, 21, 105)

print("Treinamento:", X_train.shape)
print("Validação:", X_val.shape)
print("Teste:", X_test.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Configuração para reprodutibilidade
tf.keras.utils.set_random_seed(SEED)

print(f"TensorFlow: {tf.__version__}")
```

```python
# Dimensões da entrada
num_features = X_train.shape[2]
win_size = X_train.shape[1]

# Construção da CNN-1D
model_cnn1d = keras.Sequential([
    layers.Input(shape=(win_size, num_features)),

    # Primeira camada convolucional
    layers.Conv1D(
        filters=32,
        kernel_size=3,
        strides=2,
        padding="valid",
        activation="relu"
    ),

    # Segunda camada convolucional
    layers.Conv1D(
        filters=64,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    # Redução das dimensões temporais
    layers.GlobalAveragePooling1D(),

    # Camada totalmente conectada
    layers.Dense(64, activation="relu"),

    # Saída para classificação binária
    layers.Dense(1, activation="sigmoid")

], name="CNN1D_LeagueOfLegends")

model_cnn1d.summary()
```

```python
# Conversão dos rótulos para classificação binária
y_train = (y_train == 200).astype("float32")
y_val = (y_val == 200).astype("float32")
y_test = (y_test == 200).astype("float32")

print("Distribuição dos rótulos de treinamento:")
print(pd.Series(y_train).value_counts().sort_index())
```

```python
# Compilação da CNN-1D
model_cnn1d.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.002),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Interrompe o treinamento caso a validação não melhore
early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)
```

```python
# Treinamento da CNN-1D
history_cnn1d = model_cnn1d.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=15,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Curva de Loss
axes[0].plot(
    history_cnn1d.history["loss"],
    label="Treinamento"
)

axes[0].plot(
    history_cnn1d.history["val_loss"],
    label="Validação"
)

axes[0].set_title("Loss por época")
axes[0].set_xlabel("Época")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.4)

# Curva de Accuracy
axes[1].plot(
    history_cnn1d.history["accuracy"],
    label="Treinamento"
)

axes[1].plot(
    history_cnn1d.history["val_accuracy"],
    label="Validação"
)

axes[1].set_title("Accuracy por época")
axes[1].set_xlabel("Época")
axes[1].set_ylabel("Accuracy")
axes[1].legend()
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.suptitle("Curvas de Aprendizado - CNN-1D")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

# Probabilidades previstas pela CNN
y_val_prob = model_cnn1d.predict(X_val)

# Conversão das probabilidades em classes (limiar 0.5)
y_val_pred = (y_val_prob >= 0.5).astype(int).ravel()

# Cálculo das métricas
accuracy = accuracy_score(y_val, y_val_pred)
f1_macro = f1_score(y_val, y_val_pred, average="macro")
balanced_acc = balanced_accuracy_score(y_val, y_val_pred)

print(f"Accuracy:          {accuracy:.4f}")
print(f"F1 Macro:          {f1_macro:.4f}")
print(f"Balanced Accuracy: {balanced_acc:.4f}")

print("\nMatriz de confusão:")
print(confusion_matrix(y_val, y_val_pred))
```

```python
# Probabilidades previstas pela CNN no conjunto de teste
y_test_prob = model_cnn1d.predict(X_test)

# Conversão das probabilidades em classes (limiar 0.5)
y_test_pred = (y_test_prob >= 0.5).astype(int).ravel()

# Cálculo das métricas
accuracy_test = accuracy_score(y_test, y_test_pred)
f1_macro_test = f1_score(y_test, y_test_pred, average="macro")
balanced_acc_test = balanced_accuracy_score(y_test, y_test_pred)

print("=" * 50)
print("AVALIAÇÃO - CNN-1D (CONJUNTO DE TESTE)")
print("=" * 50)

print(f"Accuracy:          {accuracy_test:.4f}")
print(f"F1 Macro:          {f1_macro_test:.4f}")
print(f"Balanced Accuracy: {balanced_acc_test:.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# Calcula a matriz de confusão
cm = confusion_matrix(y_val, y_val_pred)

# Visualização
plt.figure(figsize=(6, 5))

plt.imshow(cm, interpolation="nearest", cmap="Blues")
plt.title("Matriz de Confusão - CNN-1D (Validação)")
plt.colorbar()

# Rótulos dos eixos
classes = ["Azul (100)", "Vermelha (200)"]

plt.xticks([0, 1], classes)
plt.yticks([0, 1], classes)

plt.xlabel("Classe prevista")
plt.ylabel("Classe real")

# Valores dentro das células
for i in range(2):
    for j in range(2):
        plt.text(
            j, i, str(cm[i, j]),
            ha="center",
            va="center",
            color="black",
            fontsize=13
        )

plt.tight_layout()
plt.show()
```

## Comparação entre os modelos

Após o treinamento e a avaliação da CNN-1D, seus resultados serão comparados com os modelos desenvolvidos anteriormente: Baseline (classificador ingênuo), Random Forest e XGBoost.

O objetivo dessa comparação é analisar o desempenho da arquitetura convolucional na classificação dos vencedores das partidas, verificando se a capacidade de extração de padrões temporais da CNN-1D proporciona resultados superiores aos modelos tradicionais de aprendizado de máquina.

Para garantir uma comparação consistente, serão utilizadas as mesmas métricas de avaliação: Accuracy, F1 Macro e Balanced Accuracy. Os resultados serão organizados em tabelas comparativas, permitindo identificar as diferenças de desempenho entre as abordagens.

Inicialmente, a comparação será realizada utilizando os resultados do conjunto de validação. Posteriormente, o modelo selecionado será avaliado no conjunto de teste, reservado para estimar sua capacidade de generalização em partidas não utilizadas durante o treinamento.

```python
import pandas as pd

# Resultados do Baseline
resultados_baseline = {
    "Accuracy": 0.5347,
    "F1 Macro": 0.3484,
    "Balanced Accuracy": 0.5000
}

# Resultados do Random Forest
resultados_rf = {
    "validacao": {
        "Accuracy": 0.7753,
        "F1 Macro": 0.7726,
        "Balanced Accuracy": 0.7718
    },
    "teste": {
        "Accuracy": 0.7507,
        "F1 Macro": 0.7493,
        "Balanced Accuracy": 0.7492
    }
}

# Resultados do XGBoost
resultados_xgb = {
    "validacao": {
        "Accuracy": 0.7693,
        "F1 Macro": 0.7665,
        "Balanced Accuracy": 0.7657
    },
    "teste": {
        "Accuracy": 0.7547,
        "F1 Macro": 0.7529,
        "Balanced Accuracy": 0.7525
    }
}

# Resultados da CNN-1D
resultados_cnn = {
    "validacao": {
        "Accuracy": accuracy,
        "F1 Macro": f1_macro,
        "Balanced Accuracy": balanced_acc
    },
    "teste": {
        "Accuracy": accuracy_test,
        "F1 Macro": f1_macro_test,
        "Balanced Accuracy": balanced_acc_test
    }
}

print("Resultados dos quatro modelos armazenados!")
```

```python
# Resultados de validação
dados_validacao = {
    "Baseline": resultados_baseline,
    "Random Forest": resultados_rf["validacao"],
    "XGBoost": resultados_xgb["validacao"],
    "CNN-1D": resultados_cnn["validacao"]
}

df_validacao = pd.DataFrame(dados_validacao).T

# Resultados de teste
dados_teste = {
    "Baseline": resultados_baseline,
    "Random Forest": resultados_rf["teste"],
    "XGBoost": resultados_xgb["teste"],
    "CNN-1D": resultados_cnn["teste"]
}

df_teste = pd.DataFrame(dados_teste).T

print("=" * 60)
print("COMPARAÇÃO DOS MODELOS - VALIDAÇÃO")
print("=" * 60)
display(df_validacao.round(4))

print("\n" + "=" * 60)
print("COMPARAÇÃO DOS MODELOS - TESTE")
print("=" * 60)
display(df_teste.round(4))

# Identifica o melhor modelo em cada conjunto
melhor_validacao = df_validacao["F1 Macro"].idxmax()
melhor_teste = df_teste["F1 Macro"].idxmax()

print("\nMINI-RELATÓRIO COMPARATIVO")
print("-" * 60)

print(
    f"Melhor modelo na validação: {melhor_validacao} "
    f"(F1 Macro: {df_validacao.loc[melhor_validacao, 'F1 Macro']:.4f})"
)

print(
    f"Melhor modelo no teste: {melhor_teste} "
    f"(F1 Macro: {df_teste.loc[melhor_teste, 'F1 Macro']:.4f})"
)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Métricas
metricas = ["Accuracy", "F1 Macro", "Balanced Accuracy"]

modelos = df_teste.index.tolist()

x = np.arange(len(metricas))
largura = 0.2

plt.figure(figsize=(11, 6))

for i, modelo in enumerate(modelos):
    valores = df_teste.loc[modelo, metricas].values

    plt.bar(
        x + (i - 1.5) * largura,
        valores,
        largura,
        label=modelo
    )

plt.title("Comparação de Desempenho dos Modelos - Teste")
plt.ylabel("Pontuação")
plt.xlabel("Métricas")
plt.xticks(x, metricas)
plt.ylim(0, 1)
plt.legend(title="Modelos")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
